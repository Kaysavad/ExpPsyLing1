Jeg har i linje 11 defineret word_len til at være længden af ordet, der er defineret som word. Så har jeg har printet det i linje 13 og fået resultatet 8. Jeg har i linje 15 bedt om at få én streg pr. bogstav og brugt join-metoden til at få mellemrum mellem stregerne.

```python
def indexes_of(g: str, w: str) -> [int]:
    matches = []
    for index in range(word_len): # range(word_len) == [0,1,2,3,4,5,6,7] 
        # w er word, word er "flagpole"
        # index er bogstavets placering i ordet, 1 i anden iteration
        # w[index] er "l" i anden iteration
        match = w[index] == g
        if match: 
            matches.append(index)
    return matches

def reveal(g: str, w: str, l: str) -> str:
    indexes = indexes_of(g, w)
    for index in indexes:
        line_list = list(l)
        line_list[index] = g
        l = "".join(line_list)
    print(' '.join(l))
    return l
```
